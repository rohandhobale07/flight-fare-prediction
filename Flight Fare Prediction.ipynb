```python
from sklearn.model_selection import RandomizedSearchCV

cb_param_grid = {
    'iterations': [200, 400, 600, 800],
    'depth': [4, 6, 8, 10],
    'learning_rate': [0.01, 0.03, 0.05, 0.1],
    'l2_leaf_reg': [1, 3, 5, 7]
}

cb_base = CatBoostRegressor(verbose=0, random_state=42)
cb_search = RandomizedSearchCV(
    estimator=cb_base,
    param_distributions=cb_param_grid,
    n_iter=20,
    scoring='r2',
    cv=3,
    random_state=42,
    n_jobs=-1
)
cb_search.fit(x_train, y_train)
print('Best params:', cb_search.best_params_)
```

```python
best_cbr = cb_search.best_estimator_
tuned_y_pred = best_cbr.predict(x_test)
tuned_cbr_acc = r2_score(y_test, tuned_y_pred)

print(f'Default CatBoost test R2:  {cbr_acc:.4f}')
print(f'Tuned   CatBoost test R2:  {tuned_cbr_acc:.4f}')
print(f'Improvement: {(tuned_cbr_acc - cbr_acc) * 100:.2f} percentage points')
```
